```python
import xmltodict
import glob
import json
import time
import datetime as dt
import os
import pandas as pd

final_array = []
xml_classif_dict = {}
orpha_dict = {}
inactiv_dict = {}
inactiv_codes = ["513", "8208", "8225", "8449"]

# Local path to be changed on implementation
base_path = "./output_files/"
```

```python
# Main function, recurses on nodes to extract the informations into 'xml_classif_dict'
# Uses the previously built 'orpha_dict' as well
def parcours_prof(node):
    if "Disorder" in node: #regular node
        current = { }
        children = []
        orpha = node["Disorder"]["OrphaCode"]
        name = node["Disorder"]["Name"]["#text"]
        topo = node["Disorder"]["DisorderType"]["Name"]["#text"] 
        linkedData = orpha_dict[orpha]
        classifLvl = linkedData["classifLvl"]
        flag = linkedData["FlagValue"]
        classif_lvl = linkedData["classifLvl"]
        text = f"{name} [{classifLvl} - {topo}] ORPHA:{orpha}"
        li_attr = {
            "flag" : flag, 
            "orphaCode" : orpha
            }           
    else:
        if 'ClassificationNodeChildList' in node: #235 edge case
            return parcours_prof(node["ClassificationNodeChildList"]["ClassificationNode"])
        else: #first node
            return parcours_prof(node["ClassificationNodeRootList"]["ClassificationNode"])
        
    if node['ClassificationNodeChildList']['@count'] == '0': #leaf
        current.update({"icon" : "jstree-file"})
    else: #node
        if node['ClassificationNodeChildList']['@count'] == '1':
            children.append(parcours_prof(node['ClassificationNodeChildList']['ClassificationNode']))
        else : 
            for child in node['ClassificationNodeChildList']["ClassificationNode"]:
                if not isinstance(child, str):
                    children.append(parcours_prof(child))
    current.update({
                "text" : text, 
                "li_attr" : li_attr,
                "children" : children
                })     
    if "ExternalReferenceList" in linkedData and len(linkedData["ExternalReferenceList"]) > 0: #mapping
        mapping_children = []
        for child in linkedData["ExternalReferenceList"]:
            mapping_children.append({"text" : f"{child['Source']} : {child['Reference']} ({child['Relation']})",
                                    "icon" : "R",
                                    "li_attr" : {
                                        "source" : child['Source']
                                        }
                            })
            li_attr.update({child['Source'] : child['Reference']})
        #children rempli de ref + potentiellement de noeuds enfants
        #si current est une leaf, besoin de créer un mapping node
        mapping_node = {
                    "text" : "Mappings", 
                    "icon" : "R", 
                    "children" : mapping_children,
                    "li_attr" : {
                        "references" : "ref_node"}
                        }
        children.append(mapping_node)
        children = list(reversed(children))
        current.update({
            "children" : children, 
            "text" : text, 
            "li_attr" : li_attr 
            })
    return current
```

```python
## Iterates on node to extract classifications informations and store them in 'orpha_dict' obj
def nodes_extraction(dico):
    count = dico["JDBOR"]["DisorderList"]["@count"]
    for disorder in dico["JDBOR"]["DisorderList"]["Disorder"]:
        orpha = disorder["OrphaCode"]
        if len(count) == 5: #nomenclature
            flag = disorder["FlagValue"]
            classif_lvl = disorder["ClassificationLevel"]["Name"]["#text"]
            orpha_dict[orpha] = {"FlagValue" : flag, 
                                 "classifLvl" : classif_lvl}
            if (flag in inactiv_codes):
                if(flag != "513" and "DisorderDisorderAssociationList" in disorder and disorder.get("DisorderDisorderAssociationList")["@count"] != "0"):
                    new_orpha = disorder["DisorderDisorderAssociationList"]["DisorderDisorderAssociation"]["TargetDisorder"]["OrphaCode"]
                    flag = flag + " " + new_orpha
                inactiv_dict.update({orpha : flag})
        else: #mapping
            if not 'ExternalReferenceList' in orpha_dict[orpha]:
                orpha_dict[orpha].update({"ExternalReferenceList" : []})
            refs = disorder["ExternalReferenceList"]["ExternalReference"]
            if not isinstance(refs, list):
                refs = [refs]
            for dis in refs:
                relation = dis["DisorderMappingRelation"]["Name"]["#text"].split(" ")[0]
                if "ICD" in dis["Source"]:
                    relation += ") (" + dis["DisorderMappingICDRelation"]["Name"]["#text"].split(" (")[0] 
                obj = {"Source" : dis["Source"], 
                       "Reference" : dis["Reference"], 
                       "Relation" : relation}
                orpha_dict[orpha]["ExternalReferenceList"].append(obj)
            if orpha in snomed:
                obj = {"Source" : "SNOMED-CT",
                       "Reference" : snomed[orpha],
                       "Relation" : "E"}
                snomed.pop(orpha)
                orpha_dict[orpha]["ExternalReferenceList"].append(obj)
```

```python
            
### MAIN TASK ###
## SNOMED dictionary built from official mapping ##

df = pd.read_excel(base_path + "ORPHA-SNOMEDCT_Mapping_File_production.xlsx", skiprows=[1, 2])
snomed = df.astype(str).set_index(df.columns[0])["Unnamed: 2"].to_dict()

## Builds 'orpha_dict' : orphacode -> mappings, flag, ClassifLevel dans orpha_dict

for path_file in ['ORPHAnomenclature_en.xml', 'ORPHA_ICD10_mapping_en.xml', 'ORPHA_ICD11_mapping_en.xml', 'ORPHA_OMIM_mapping_en.xml']:
    with open(base_path + "Orphanet_Nomenclature_Pack_en/" + path_file, "r", encoding="ISO-8859-1") as file:
        nodes_extraction(xmltodict.parse(file.read()))

## Builds xml_classif_dict : path -> classif + Date in xml_classif_dict for easier iteration

date = False
for path_file in sorted(glob.glob(base_path + "Orphanet_Nomenclature_Pack_en/Classifications/*.xml")):
    with open(path_file, "r", encoding="ISO-8859-1") as classif:
        xml_classif_dict.update({path_file : xmltodict.parse(classif.read())})
        if not date:
            stamp = dt.datetime.fromtimestamp(os.path.getctime(path_file)) #run sur quoi ?
            date = stamp.strftime("%Y-%m-%d")
with open(base_path + "creation_date.txt", "w") as date_file:
    date_file.write(date)

## Main Loop
# xml_classif_dict -> parcours prof on chaque classif and add to final_array 
# orpha_dict is added during parc_prof 

for k,v in xml_classif_dict.items():
    classif = v["JDBOR"]["ClassificationList"]["Classification"]
    classif_object = parcours_prof(classif)
    # Only active codes found during classification node extraction are considered here
    final_array.append(classif_object)
with open(base_path + "/arborData_000.json", "w") as output_file:
    output_file.write(json.dumps(final_array))

# Output a file containing all inactiv codes 
with open(base_path + "/inactiv_codes.json", "w") as output_file:
    output_file.write(json.dumps(inactiv_dict))
```
